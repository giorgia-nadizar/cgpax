```python
import pandas as pd
```

```python
df = pd.read_csv("../datasets/FeynmanEquations.csv")
df.head()
```

```python
n_vars = 5
df_tmp = df[df["# variables"] == n_vars]
variables_list = []

for i, row in df_tmp.iterrows():
    these_variables = sorted([row[f"v{i + 1}_name"] for i in range(n_vars)], key=str.lower)
    # these_variables = [row[f"v{i + 1}_name"] for i in range(n_vars)]
    variables_list.append(these_variables)

for i, var_list in enumerate(variables_list):
        print(i, var_list)
```
